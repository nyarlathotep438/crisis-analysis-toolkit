```python
import numpy as np
import pandas as pd

np.random.seed(42)  # 保证复现性

def generate_test_data(n=500):
    # 正常数据：均值 50，标准差 5
    normal = np.random.normal(loc=50, scale=5, size=n)

    # 异常点：随机挑选 10 个位置，加入极端值
    anomaly_indices = np.random.choice(n, size=10, replace=False)
    normal[anomaly_indices] += np.random.choice([30, -30, 50, -50], size=10)

    # 异常段：连续漂移（模拟传感器偏移）
    drift_start = np.random.randint(100, 300)
    drift_length = 30
    normal[drift_start: drift_start + drift_length] += np.linspace(0, 40, drift_length)

    # 缺失值
    missing_indices = np.random.choice(n, size=8, replace=False)
    normal[missing_indices] = np.nan

    # 构造 DataFrame
    df = pd.DataFrame({
        "timestamp": pd.date_range("2024-01-01", periods=n, freq="T"),
        "value": normal
    })

    # 重复行（数据质量异常）
    df = pd.concat([df, df.iloc[[50, 120]]], ignore_index=True)

    return df

df_test = generate_test_data()
df_test.head()
```
